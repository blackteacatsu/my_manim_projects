## Build background canvas

```python
%%manim -qm HydroViewerBackground

class HydroViewerBackground(Scene):
    def construct(self):
        # AmazonHydroViewer brutalist color palette
        WHITE = "#ffffff"
        BLACK = "#000000"
        LIGHT_GRAY = "#f5f5f5"
        
        # Border width from the CSS
        BORDER_WIDTH = 4  # 2px scaled up for Manim
        
        # Set white background for brutalist look
        self.camera.background_color = BLACK
        
        # Create bold black border frame
        border = Rectangle(
            width=config.frame_width - 0.4,
            height=config.frame_height - 0.4,
            stroke_color=WHITE,
            stroke_width=BORDER_WIDTH,
            fill_opacity=0
        )
        
        # Add light gray accent panel at top (header area)
        header = Rectangle(
            width=config.frame_width - 0.4,
            height=1.5,
            fill_color=BLACK,
            fill_opacity=1,
            stroke_color=WHITE,
            stroke_width=BORDER_WIDTH
        ).to_edge(UP, buff=0.2)
        
        # Create offset shadow effect for header (brutalist micro-interaction)
        header_shadow = Rectangle(
            width=config.frame_width - 0.4,
            height=1.5,
            fill_color=BLACK,
            fill_opacity=1,
            stroke_width=0
        ).to_edge(UP, buff=0.2).shift(RIGHT * 0.05 + DOWN * 0.05)
        
        # Add title with Space Grotesk font (bold, uppercase)
        title = Text(
            "AMAZONHYDROVIEWER",
            font="Space Grotesk",  # Using the actual Space Grotesk font!
            font_size=48,
            color=WHITE,
            #weight=REGULAR
        ).move_to(header.get_center())
        
        # Add content area with light gray background
        content_bg = Rectangle(
            width=config.frame_width - 0.8,
            height=config.frame_height - 2.5,
            fill_color=WHITE,
            fill_opacity=0.3,
            stroke_width=0
        ).shift(DOWN * 0.3)
        
        # Add decorative elements with offset shadow
        card1 = Rectangle(
            width=5,
            height=2,
            fill_color=BLACK,
            fill_opacity=1,
            stroke_color=WHITE,
            stroke_width=BORDER_WIDTH
        ).shift(LEFT * 3 + DOWN * 1)
        
        card1_shadow = Rectangle(
            width=5,
            height=2,
            fill_color=WHITE,
            fill_opacity=1,
            stroke_width=0
        ).shift(LEFT * 3 + DOWN * 1 + RIGHT * 0.08 + DOWN * 0.08)
        
        card2 = Rectangle(
            width=5,
            height=2,
            fill_color=BLACK,
            fill_opacity=1,
            stroke_color=WHITE,
            stroke_width=BORDER_WIDTH
        ).shift(RIGHT * 3 + DOWN * 1)
        
        card2_shadow = Rectangle(
            width=5,
            height=2,
            fill_color=WHITE,
            fill_opacity=1,
            stroke_width=0
        ).shift(RIGHT * 3 + DOWN * 1 + RIGHT * 0.08 + DOWN * 0.08)
        
        # Add subtitle with Space Grotesk
        subtitle = Text(
            "GIS Day 2024",
            font="Space Grotesk",
            font_size=24,
            color=WHITE,
            #weight=REGULAR
        ).next_to(header, DOWN, buff=0.5)
        
        # Assemble scene
        self.add(content_bg)
        self.add(card1_shadow, card1)
        self.add(card2_shadow, card2)
        self.add(header_shadow, header, border)
        
        # Animate title in
        self.play(Write(title))
        self.play(
            FadeIn(subtitle, shift=DOWN * 0.1),
            run_time=1
        )
        self.wait(2)
```

```python
%%manim -qm XarrayOpenData

class XarrayOpenData(ThreeDScene):
    def construct(self):
        # Intro with text
        greetings = Text("Hi there!", font="Space Grotesk", font_size=36, color=WHITE)
        title = Text("What's inside a NetCDF file?", font="Space Grotesk", font_size=36, color=WHITE)
        
        self.play(Write(greetings))
        self.wait(1)
        title.to_corner(UP + LEFT)
        self.play(Transform(greetings, title))
        self.wait(1)
        
        # Fade out title and show NetCDF visualization
        self.play(FadeOut(greetings))
        
        # Create initial 2D plane (representing a spatial grid)
        plane_color = BLUE
        grid_size = 3
        
        # Initial 2D grid
        initial_plane = Square(
            side_length=grid_size,
            fill_color=plane_color,
            fill_opacity=0.3,
            stroke_color=WHITE,
            stroke_width=2
        )
        
        # Add grid lines to show it's a data grid
        grid_lines = VGroup()
        for i in range(-1, 2):
            # Horizontal lines
            h_line = Line(
                start=[-grid_size/2, i * grid_size/3, 0],
                end=[grid_size/2, i * grid_size/3, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=0.5
            )
            grid_lines.add(h_line)
            
            # Vertical lines
            v_line = Line(
                start=[i * grid_size/3, -grid_size/2, 0],
                end=[i * grid_size/3, grid_size/2, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=0.5
            )
            grid_lines.add(v_line)
        
        plane_2d = VGroup(initial_plane, grid_lines)
        
        # Labels for 2D view
        x_label_2d = Text("X (Longitude)", font="Space Grotesk", font_size=20, color=WHITE).next_to(initial_plane, DOWN)
        y_label_2d = Text("Y (Latitude)", font="Space Grotesk", font_size=20, color=WHITE).next_to(initial_plane, LEFT).rotate(PI/2)
        
        desc_2d = Text("2D Spatial Grid", font="Space Grotesk", font_size=24, color=WHITE).to_edge(UP)
        
        # Show 2D plane
        self.play(Create(plane_2d), Write(desc_2d))
        self.play(Write(x_label_2d), Write(y_label_2d))
        self.wait(2)
        
        # Transition to 3D
        transition_text = Text("Adding Time Dimension...", font="Space Grotesk", font_size=24, color=WHITE).to_edge(UP)
        self.play(Transform(desc_2d, transition_text))
        
        # Remove 2D labels
        self.play(FadeOut(x_label_2d), FadeOut(y_label_2d))
        
        # Set up 3D camera
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES)
        
        # Convert 2D plane to 3D
        plane_3d = Square(
            side_length=grid_size,
            fill_color=plane_color,
            fill_opacity=0.3,
            stroke_color=WHITE,
            stroke_width=2
        )
        grid_3d = VGroup()
        for i in range(-1, 2):
            h_line = Line(
                start=[-grid_size/2, i * grid_size/3, 0],
                end=[grid_size/2, i * grid_size/3, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=0.5
            )
            grid_3d.add(h_line)
            v_line = Line(
                start=[i * grid_size/3, -grid_size/2, 0],
                end=[i * grid_size/3, grid_size/2, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=0.5
            )
            grid_3d.add(v_line)
        
        first_layer = VGroup(plane_3d, grid_3d)
        
        # Rotate camera to show 3D
        self.play(
            ReplacementTransform(plane_2d, first_layer)
        )
        self.begin_ambient_camera_rotation(rate=0.1)
        self.wait(2)
        
        # Create additional time layers
        num_layers = 6
        layers = VGroup(first_layer)
        
        for i in range(1, num_layers):
            # Create new plane
            new_plane = Square(
                side_length=grid_size,
                fill_color=plane_color,
                fill_opacity=0.3,
                stroke_color=WHITE,
                stroke_width=2
            ).shift(UP * i * 0.4)
            
            new_grid = VGroup()
            for j in range(-1, 2):
                h_line = Line(
                    start=[-grid_size/2, j * grid_size/3, i * 0.4],
                    end=[grid_size/2, j * grid_size/3, i * 0.4],
                    stroke_color=WHITE,
                    stroke_width=1,
                    stroke_opacity=0.5
                )
                new_grid.add(h_line)
                v_line = Line(
                    start=[j * grid_size/3, -grid_size/2, i * 0.4],
                    end=[j * grid_size/3, grid_size/2, i * 0.4],
                    stroke_color=WHITE,
                    stroke_width=1,
                    stroke_opacity=0.5
                )
                new_grid.add(v_line)
            
            layer = VGroup(new_plane, new_grid)
            layers.add(layer)
        
        # Stack the layers
        self.stop_ambient_camera_rotation()
        self.play(
            *[FadeIn(layers[i], shift=UP * 0.2) for i in range(1, num_layers)],
            run_time=2
        )
        
        # Add 3D axis labels
        time_arrow = Arrow3D(
            start=[2, 0, 0],
            end=[2, 0, 2.5],
            color=YELLOW,
            thickness=0.02
        )
        
        time_label = Text("Time", font="Space Grotesk", font_size=20, color=YELLOW)
        time_label.rotate(PI/2, axis=RIGHT).rotate(PI/2, axis=OUT)
        time_label.next_to([2, 0, 2.5], OUT)
        
        self.play(Create(time_arrow), Write(time_label))
        
        # Final rotation
        self.move_camera(phi=60 * DEGREES, theta=-60 * DEGREES, run_time=3)
        self.wait(2)
```
