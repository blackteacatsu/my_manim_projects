```python
%%manim -qm XarrayOpenData

class XarrayOpenData(Scene):
    def construct(self):
        # Try Text with explicit parameters like the first scene
        hi = Text("Hi there!", font="Space Grotesk", font_size=36, color=WHITE)
        title = Text("Is this you when opening a NetCDF file?", font="Space Grotesk", font_size=36, color=WHITE)
        question = Text("What if there is an easier way?", font="Space Grotesk", font_size=36, color=WHITE)
        cursor = Rectangle(
            color = GREY_A,
            fill_color = GREY_A,
            fill_opacity = 1.0,
            height = 0.5,
            width = 0.2,
        ).move_to(hi[0])
        code = '''import xarray as xr

FILE_PATH = "/mnt/vast/usr/ldas_forecast_data/2025_09.nc"
ds=xr.open_dataset(FILE_PATH, engine="netcdf4")
da = ds['Qair_tavg'].isel(time=0)
ds.close()

import matplotlib.pyplot as plt
da.plot(figsize=(12, 6), cmap='RdBu_r', vmin=-1, vmax=1)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
'''
        rendered_code = Code(
            code_string=code, tab_width=4, background="window",
            language="python", paragraph_config=dict(font="Monospace")
        )
        self.play(TypeWithCursor(hi, cursor))
        self.play(Blink(cursor, blinks=2))
        title.to_corner(UP + LEFT)
        self.play(Transform(hi, title))
        self.play(Write(rendered_code))
        self.wait(3)
        cursor = Rectangle(
                    color = GREY_A,
                    fill_color = GREY_A,
                    fill_opacity = 1.0,
                    height = 0.5,
                    width = 0.2,
                ).move_to(question[0])
        self.play(Unwrite(hi), Unwrite(rendered_code, reverse=False))
        self.play(TypeWithCursor(question, cursor))
        self.play(Blink(cursor, blinks=2))
        self.play(FadeOut(question, shift=UP))
```

```python
%%manim -qm NetCDFStructure

class NetCDFStructure(ThreeDScene):
    def construct(self):
        # Start with camera pointing straight down (top-down view)
        self.set_camera_orientation(phi=0 * DEGREES, theta= -90 * DEGREES, zoom=0.95)
        title = Text("NetCDF Structure At a Glance", font="Space Grotesk", font_size=36, color=WHITE)
        title.to_edge(UP)
        # Show from top
        self.add_fixed_in_frame_mobjects(title)
        self.play(Write(title))
        
        # Create first 2D plane (looks like a square from top)
        grid_size = 6
        plane_color = BLUE
        
        # Create plane with grid lines
        first_plane = Square(
            side_length=grid_size,
            fill_color=plane_color,
            fill_opacity=0.4,
            stroke_color=WHITE,
            stroke_width=2
        )
        
        grid = NumberPlane(
            x_range=(0, 8, 1),
            y_range=(0, 8, 1),
            #axis_config={"show_axis": True},
        )
        
        # Add grid lines
        # threed_grid = ThreeDAxes(
        #     x_range=(0, 8, 1),
        #     y_range=(0, 8, 1),
        #     z_range=(0, 8, 1),
        #     #axis_config={"show": False},
        # )
        #y_label = grid.get_y_axis_label("x", edge=LEFT, direction=LEFT, buff=0.4)
        #x_label = grid.get_x_axis_label("y", edge=DOWN, direction=DOWN, buff=0.4)

        # Title text
        grid_title = Tex("This is a grid")
        grid_title.scale(1.5)
        self.add(grid, grid_title)
        self.play(
            FadeIn(grid_title, shift=DOWN),
            Create(grid, run_time=3, lag_ratio=0.1),
            #Create(threed_grid)
        )
        self.wait()


        grid_lines = VGroup()
        num_lines = 6
        for i in range(num_lines + 1):
            # Horizontal lines
            h_line = Line(
                start=[-grid_size/2, -grid_size/2 + i * grid_size/num_lines, 0],
                end=[grid_size/2, -grid_size/2 + i * grid_size/num_lines, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=1
            )
            grid_lines.add(h_line)
            
            # Vertical lines
            v_line = Line(
                start=[-grid_size/2 + i * grid_size/num_lines, -grid_size/2, 0],
                end=[-grid_size/2 + i * grid_size/num_lines, grid_size/2, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=1
            )
            grid_lines.add(v_line)
        
        plane_2d = VGroup(first_plane, grid_lines)

        
        self.play(DrawBorderThenFill(plane_2d), Unwrite(grid_title))
        self.wait(1)
        
        # Pan camera to show it's actually 3D
        self.move_camera(phi=45 * DEGREES, theta=-45 * DEGREES, run_time=3, zoom=0.5)
        self.wait(1)
        
        # Update title
        new_title = Text("Add a new entry for each time step", font="Space Grotesk", font_size=36, color=WHITE).to_edge(UP)
        #self.remove_fixed_in_frame_mobjects(title)
        self.add_fixed_in_frame_mobjects(new_title)
        self.play(FadeOut(title, shift=UP), Write(new_title))
        
        # Create and stack additional layers
        num_layers = 6
        layer_spacing = 1
        all_layers = VGroup(plane_2d)
        
        for i in range(1, num_layers):
            # Create 3D surface for each layer
            def make_plane_func(z_val):
                return lambda u, v: np.array([u, v, z_val])
            
            # Create new plane
            new_plane = Surface(
                make_plane_func(i * layer_spacing),
                u_range=[-grid_size/2, grid_size/2],
                v_range=[-grid_size/2, grid_size/2],
                resolution=(1, 1),
                fill_color=plane_color,
                fill_opacity=0.6,
                stroke_color=WHITE,
                stroke_width=2,
            )
            
            # Create grid for this layer
            new_grid = VGroup()
            for j in range(num_lines + 1):
                h_line = Line(
                    start=[-grid_size/2, -grid_size/2 + j * grid_size/num_lines, i * layer_spacing],
                    end=[grid_size/2, -grid_size/2 + j * grid_size/num_lines, i * layer_spacing],
                    stroke_color=WHITE,
                    stroke_width=0.6,
                    stroke_opacity=1
                )
                new_grid.add(h_line)
                
                v_line = Line(
                    start=[-grid_size/2 + j * grid_size/num_lines, -grid_size/2, i * layer_spacing],
                    end=[-grid_size/2 + j * grid_size/num_lines, grid_size/2, i * layer_spacing],
                    stroke_color=WHITE,
                    stroke_width=1,
                    stroke_opacity=0.6
                )
                new_grid.add(v_line)
            
            layer = VGroup(new_plane,new_grid)
            all_layers.add(layer)
            
            # Animate stacking
            self.play(FadeIn(layer, shift=UP * 0.3), run_time=0.5)
        
        self.move_camera(phi=60 * DEGREES, theta=-70 * DEGREES, run_time=3, zoom=0.3)
        self.begin_3dillusion_camera_rotation(rate=2)
        self.wait(PI/2)
        self.stop_3dillusion_camera_rotation()
```
