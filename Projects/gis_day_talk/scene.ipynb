```python
%%manim -qm XarrayOpenData

class XarrayOpenData(Scene):
    def construct(self):
        # Try Text with explicit parameters like the first scene
        hi = Text("Hi there!", font="Space Grotesk", font_size=36, color=WHITE)
        title = Text("Is this you when opening a NetCDF file?", font="Space Grotesk", font_size=36, color=WHITE)
        question = Text("What if there is an easier way?", font="Space Grotesk", font_size=36, color=WHITE)
        cursor = Rectangle(
            color = GREY_A,
            fill_color = GREY_A,
            fill_opacity = 1.0,
            height = 0.5,
            width = 0.2,
        ).move_to(hi[0])
        code = '''import xarray as xr

FILE_PATH = "/mnt/vast/usr/ldas_forecast_data/2025_09.nc"
ds=xr.open_dataset(FILE_PATH, engine="netcdf4")
da = ds['Qair_tavg'].isel(time=0)
ds.close()

import matplotlib.pyplot as plt
da.plot(figsize=(12, 6), cmap='RdBu_r', vmin=-1, vmax=1)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
'''
        rendered_code = Code(
            code_string=code, tab_width=4, background="window",
            language="python", paragraph_config=dict(font="Monospace")
        )
        self.play(TypeWithCursor(hi, cursor))
        self.play(Blink(cursor, blinks=2))
        title.to_corner(UP + LEFT)
        self.play(Transform(hi, title))
        self.play(Write(rendered_code))
        self.wait(3)
        cursor = Rectangle(
                    color = GREY_A,
                    fill_color = GREY_A,
                    fill_opacity = 1.0,
                    height = 0.5,
                    width = 0.2,
                ).move_to(question[0])
        self.play(Unwrite(hi), Unwrite(rendered_code, reverse=False))
        self.play(TypeWithCursor(question, cursor))
        self.play(Blink(cursor, blinks=2))
        self.play(FadeOut(question, shift=UP))
```

```python
%%manim -qm NetCDFStructure

class NetCDFStructure(ThreeDScene):
    def construct(self):
        # Start with camera pointing straight down (top-down view)
        self.set_camera_orientation(phi=0 * DEGREES, theta= -90 * DEGREES)
        title = Text("NetCDF Data Structure", font="Space Grotesk", font_size=36, color=WHITE)
        title.to_edge(UP)
        # Show from top
        self.add_fixed_in_frame_mobjects(title)
        self.play(Write(title))
        
        # Create first 2D plane (looks like a square from top)
        grid_size = 6
        plane_color = BLUE
        
        # Create plane with grid lines
        first_plane = Square(
            side_length=grid_size,
            fill_color=plane_color,
            fill_opacity=0.4,
            stroke_color=WHITE,
            stroke_width=2
        )
        
        grid = NumberPlane(
            x_range=(0, 8, 1),
            y_range=(0, 8, 1),
            #axis_config={"show_axis": True},
        )
        
        # Add grid lines
        # threed_grid = ThreeDAxes(
        #     x_range=(0, 8, 1),
        #     y_range=(0, 8, 1),
        #     z_range=(0, 8, 1),
        #     #axis_config={"show": False},
        # )
        #y_label = grid.get_y_axis_label("x", edge=LEFT, direction=LEFT, buff=0.4)
        #x_label = grid.get_x_axis_label("y", edge=DOWN, direction=DOWN, buff=0.4)

        # Title text
        grid_title = Tex("This is a grid")
        grid_title.scale(1.5)
        self.add(grid, grid_title)
        self.play(
            FadeIn(grid_title, shift=DOWN),
            Create(grid, run_time=3, lag_ratio=0.1),
            #Create(threed_grid)
        )
        self.wait()


        grid_lines = VGroup()
        num_lines = 6
        for i in range(num_lines + 1):
            # Horizontal lines
            h_line = Line(
                start=[-grid_size/2, -grid_size/2 + i * grid_size/num_lines, 0],
                end=[grid_size/2, -grid_size/2 + i * grid_size/num_lines, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=1
            )
            grid_lines.add(h_line)
            
            # Vertical lines
            v_line = Line(
                start=[-grid_size/2 + i * grid_size/num_lines, -grid_size/2, 0],
                end=[-grid_size/2 + i * grid_size/num_lines, grid_size/2, 0],
                stroke_color=WHITE,
                stroke_width=1,
                stroke_opacity=1
            )
            grid_lines.add(v_line)
        
        plane_2d = VGroup(first_plane, grid_lines)

        
        self.play(DrawBorderThenFill(plane_2d), Unwrite(grid_title))
        self.wait(1)
        
        # Pan camera to show it's actually 3D
        self.move_camera(phi=70 * DEGREES, theta=-45 * DEGREES, run_time=3, zoom=0.5)
        self.wait(1)
        
        # Update title
        new_title = Text("Stacking Time Layers", font="Space Grotesk", font_size=36, color=WHITE)
        new_title.to_edge(UP)
        #self.add_fixed_in_frame_mobjects(new_title)
        self.play(Transform(title, new_title))
        
        # Create and stack additional layers
        num_layers = 6
        layer_spacing = 1
        all_layers = VGroup(plane_2d)
        
        for i in range(1, num_layers):
            # Create new plane
            new_plane = ThreeDVMobjectSquare(
                side_length=grid_size,
                fill_color=plane_color,
                fill_opacity=0.4,
                stroke_color=WHITE,
                stroke_width=2
            ).shift(z = UP * i * layer_spacing)
            
            # Create grid for this layer
            new_grid = VGroup()
            for j in range(num_lines + 1):
                h_line = Line(
                    start=[-grid_size/2, -grid_size/2 + j * grid_size/num_lines, i * layer_spacing],
                    end=[grid_size/2, -grid_size/2 + j * grid_size/num_lines, i * layer_spacing],
                    stroke_color=WHITE,
                    stroke_width=1,
                    stroke_opacity=0.6
                )
                new_grid.add(h_line)
                
                v_line = Line(
                    start=[-grid_size/2 + j * grid_size/num_lines, -grid_size/2, i * layer_spacing],
                    end=[-grid_size/2 + j * grid_size/num_lines, grid_size/2, i * layer_spacing],
                    stroke_color=WHITE,
                    stroke_width=1,
                    stroke_opacity=1
                )
                new_grid.add(v_line)
            
            layer = VGroup(new_plane,new_grid)#new_plane)#, new_grid)
            all_layers.add(layer)
            
            # Animate stacking
            self.play(FadeIn(layer, shift=UP * 0.3), run_time=0.5)
        
        self.wait(1)
        
        # Add time axis label
        # time_arrow = Arrow3D(
        #     start=[grid_size/2 + 1, 0, 0],
        #     end=[grid_size/2 + 1, 0, (num_layers - 1) * layer_spacing + 0.5],
        #     color=YELLOW,
        #     thickness=0.02
        # )
        
        # time_label = Text("Time", font="Space Grotesk", font_size=24, color=YELLOW)
        # time_label.rotate(PI/2, axis=RIGHT).rotate(PI/2, axis=OUT)
        # time_label.move_to([grid_size/2 + 1.5, 0, (num_layers - 1) * layer_spacing/2])
        
        #self.play(Create(time_arrow), Write(time_label))
        
        # Final camera rotation
        self.move_camera(phi=60 * DEGREES, theta=-60 * DEGREES, run_time=3)
        self.wait(2)
```

```python
%%manim -qm TercileProbability

class TercileProbability(ThreeDScene):
    def construct(self):
        # Start in 3D with historical hindcast layers
        self.set_camera_orientation(phi=60 * DEGREES, theta=-45 * DEGREES)
        
        # Title
        title = Text("Calculating Tercile Probabilities", font="Space Grotesk", font_size=32, color=WHITE)
        title.to_edge(UP)
        self.add_fixed_in_frame_mobjects(title)
        self.play(Write(title))
        self.wait(1)
        
        # Step 1: Show hindcast layers (1982-2017)
        subtitle = Text("Historical Data (1982-2017)", font="Space Grotesk", font_size=24, color=WHITE)
        subtitle.next_to(title, DOWN)
        self.add_fixed_in_frame_mobjects(subtitle)
        self.play(Write(subtitle))
        
        # Create simplified hindcast layers with varying colors (representing different values)
        num_years = 36  # 1982-2017
        layer_height = 0.1
        hindcast_layers = VGroup()
        
        # Simulate different values with color gradient
        for i in range(num_years):
            value = i / num_years  # 0 to 1
            color = interpolate_color(BLUE, RED, value)
            
            layer = Rectangle(
                width=3,
                height=layer_height,
                fill_color=color,
                fill_opacity=0.8,
                stroke_color=WHITE,
                stroke_width=1
            ).shift(UP * i * layer_height * 1.2 - UP * 2)
            hindcast_layers.add(layer)
        
        self.play(LaggedStart(*[FadeIn(layer, shift=UP * 0.1) for layer in hindcast_layers], lag_ratio=0.05))
        self.wait(1)
        
        # Step 2: Rank the data
        new_subtitle = Text("Rank from Lowest to Highest", font="Space Grotesk", font_size=24, color=WHITE)
        new_subtitle.next_to(title, DOWN)
        #self.add_fixed_in_frame_mobjects(new_subtitle)
        self.play(Transform(subtitle, new_subtitle))
        
        # Already sorted by color, just emphasize
        self.play(hindcast_layers.animate.shift(LEFT * 2))
        self.wait(1)
        
        # Step 3: Divide into terciles
        tercile_title = Text("Divide into 3 Categories", font="Space Grotesk", font_size=24, color=WHITE)
        tercile_title.next_to(title, DOWN)
        #self.add_fixed_in_frame_mobjects(tercile_title)
        self.play(Transform(subtitle, tercile_title))
        
        # Add category labels
        below_label = Text("Below\nNormal", font="Space Grotesk", font_size=16, color=BLUE)
        below_label.rotate(PI/2, axis=RIGHT).rotate(PI/2, axis=OUT)
        below_label.move_to([-4, 0, -1.5])
        
        near_label = Text("Near\nNormal", font="Space Grotesk", font_size=16, color=GRAY)
        near_label.rotate(PI/2, axis=RIGHT).rotate(PI/2, axis=OUT)
        near_label.move_to([-4, 0, 0])
        
        above_label = Text("Above\nNormal", font="Space Grotesk", font_size=16, color=RED)
        above_label.rotate(PI/2, axis=RIGHT).rotate(PI/2, axis=OUT)
        above_label.move_to([-4, 0, 1.5])
        
        # Draw dividing lines
        tercile_size = num_years // 3
        div_line1_y = hindcast_layers[tercile_size].get_center()[2]
        div_line2_y = hindcast_layers[2 * tercile_size].get_center()[2]
        
        div_line1 = Line(
            start=[-3.5, 0, div_line1_y],
            end=[-0.5, 0, div_line1_y],
            stroke_color=YELLOW,
            stroke_width=4
        )
        
        div_line2 = Line(
            start=[-3.5, 0, div_line2_y],
            end=[-0.5, 0, div_line2_y],
            stroke_color=YELLOW,
            stroke_width=4
        )
        
        self.play(
            Create(div_line1),
            Create(div_line2),
            Write(below_label),
            Write(near_label),
            Write(above_label)
        )
        self.wait(2)
        
        # Step 4: Bring in forecast ensembles
        forecast_title = Text("Compare Forecast Ensembles", font="Space Grotesk", font_size=24, color=WHITE)
        forecast_title.next_to(title, DOWN)
        #self.add_fixed_in_frame_mobjects(forecast_title)
        self.play(Transform(subtitle, forecast_title))
        
        # Create ensemble members (10 ensembles with different values)
        ensemble_values = [0.15, 0.45, 0.05, 0.75, 0.38, 0.32, 0.85, 0.65, 0.95, 0.52]
        ensembles = VGroup()
        
        for i, value in enumerate(ensemble_values):
            color = interpolate_color(BLUE, RED, value)
            ensemble = Rectangle(
                width=0.3,
                height=layer_height,
                fill_color=color,
                fill_opacity=0.9,
                stroke_color=YELLOW,
                stroke_width=2
            ).shift(RIGHT * 2 + UP * (i * 0.15 - 1))
            ensembles.add(ensemble)
        
        self.play(LaggedStart(*[FadeIn(e, shift=LEFT) for e in ensembles], lag_ratio=0.1))
        self.wait(1)
        
        # Step 5: Categorize ensembles
        categorize_title = Text("Categorize Each Ensemble", font="Space Grotesk", font_size=24, color=WHITE)
        categorize_title.next_to(title, DOWN)
        #self.add_fixed_in_frame_mobjects(categorize_title)
        self.play(Transform(subtitle, categorize_title))
        
        # Count categories
        below_count = sum(1 for v in ensemble_values if v < 0.33)
        near_count = sum(1 for v in ensemble_values if 0.33 <= v < 0.67)
        above_count = sum(1 for v in ensemble_values if v >= 0.67)
        
        # Move ensembles to their categories
        animations = []
        below_pos = 0
        near_pos = 0
        above_pos = 0
        
        for i, (ensemble, value) in enumerate(zip(ensembles, ensemble_values)):
            if value < 0.33:
                target = [-2, 0, -1.5 + below_pos * 0.15]
                below_pos += 1
            elif value < 0.67:
                target = [-2, 0, 0 + near_pos * 0.15]
                near_pos += 1
            else:
                target = [-2, 0, 1.5 + above_pos * 0.15]
                above_pos += 1
            animations.append(ensemble.animate.move_to(target))
        
        self.play(*animations, run_time=2)
        self.wait(1)
        
        # Step 6: Calculate probabilities
        prob_title = Text("Calculate Probabilities", font="Space Grotesk", font_size=24, color=WHITE)
        prob_title.next_to(title, DOWN)
        #self.add_fixed_in_frame_mobjects(prob_title)
        self.play(Transform(subtitle, prob_title))
        
        # Show probability text
        prob_below = Text(f"Below: {below_count}/10 = {below_count/10:.0%}", 
                         font="Space Grotesk", font_size=20, color=BLUE)
        prob_below.to_corner(DOWN + LEFT).shift(RIGHT * 0.5)
        
        prob_near = Text(f"Near: {near_count}/10 = {near_count/10:.0%}", 
                        font="Space Grotesk", font_size=20, color=GRAY)
        prob_near.next_to(prob_below, RIGHT, buff=0.5)
        
        prob_above = Text(f"Above: {above_count}/10 = {above_count/10:.0%}", 
                         font="Space Grotesk", font_size=20, color=RED)
        prob_above.next_to(prob_near, RIGHT, buff=0.5)
        
        self.add_fixed_in_frame_mobjects(prob_below, prob_near, prob_above)
        self.play(
            Write(prob_below),
            Write(prob_near),
            Write(prob_above)
        )
        self.wait(2)
        
        # Final camera rotation
        self.move_camera(phi=70 * DEGREES, theta=-60 * DEGREES, run_time=2)
        self.wait(2)
```
